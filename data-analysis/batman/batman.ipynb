```python
import numpy as np
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

with open('buildings.dat.txt') as file:
    _ = next(file)
    data = np.array([tuple(map(int, line.split())) for line in file.readlines()])

dbskan = DBSCAN(eps=300, min_samples=1)
result = dbskan.fit_predict(data)

plt.scatter(data[:, 0], data[:, 1], c=result)
plt.grid()
plt.show()
```
